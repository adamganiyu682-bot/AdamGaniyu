```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

sizes = [50, 75, 100, 125, 150,175, 200, 225, 250, 275]
prices = [120000, 170000, 220000, 270000, 320000,370000, 420000, 470000, 520000, 570000]



data = { "sizes": sizes,
         "prices":  prices
          }

df  = pd.DataFrame(data)

X = df[["sizes"]]
y = df["prices"]

model = LinearRegression()

model.fit(X,y)

prediction = model.predict([[80],[110],[140]])

print(prediction)
```
